```python
# Calculate local listing counts and rent-per-square-meter quartiles
group_columns = ["city", "town"]

grouped_prices = rent_price_check_df.groupby(
    group_columns
)["monthly_rent_per_sqm"]

rent_price_check_df["area_group_count"] = grouped_prices.transform("count")
rent_price_check_df["area_q1"] = grouped_prices.transform(
    lambda values: values.quantile(0.25)
)
rent_price_check_df["area_q3"] = grouped_prices.transform(
    lambda values: values.quantile(0.75)
)

# Calculate the interquartile range for each city-town group
rent_price_check_df["area_iqr"] = (
    rent_price_check_df["area_q3"] - rent_price_check_df["area_q1"]
)

# Flag values outside the local 1.5-IQR range
rent_price_check_df["local_price_outlier"] = (
    (rent_price_check_df["area_group_count"] >= 30) &
    (
        (rent_price_check_df["monthly_rent_per_sqm"] <
         rent_price_check_df["area_q1"] - 1.5 * rent_price_check_df["area_iqr"]) |
        (rent_price_check_df["monthly_rent_per_sqm"] >
         rent_price_check_df["area_q3"] + 1.5 * rent_price_check_df["area_iqr"])
    )
)

# Display flag counts without deleting any listings
print("Listings with a local comparison group:", (
    rent_price_check_df["area_group_count"] >= 30
).sum())

print("Listings flagged for local price review:", (
    rent_price_check_df["local_price_outlier"]
).sum())
```

## Cleaning and Record-Retention Rules — Initial Policy

### 1. Data preservation
- Keep the original dataset in `df` unchanged.
- Perform all cleaning and filtering on separate working copies.
- Never permanently delete a listing because it is unusual or flagged.
- Record a status and a reason for every listing that is reviewed or excluded from a specific analysis.

### 2. Analysis scope
- The initial analysis focuses on residential apartment rentals.
- Listings classified as monthly rentals and apartments may enter the initial rental-analysis dataset.
- Listings classified as sales or as other property types are kept in the raw dataset but excluded from the apartment-rental analysis.
- The project team should confirm whether hotel apartments and short-term rentals are in scope before including them in the student-housing price reference.

### 3. Listing type and text conflicts
- Use `price_period` and `offering_type` as structured source fields to identify the stated listing type.
- If the title or description contains clear sale-related wording while the structured fields say monthly rent, flag the listing as `REVIEW_TYPE_CONFLICT`.
- Do not automatically classify a listing as fraud based on keywords.
- Keep type-conflict listings out of the initial rental-price reference until reviewed.
- Preserve the listing and its original values in a separate review list.

### 4. Price and area validity
- Keep `price_egp` and `area_value` only as numeric values in the analysis copy.
- A missing, zero, or negative price or area makes the listing ineligible for price-per-square-meter calculations until corrected or reviewed.
- Confirm `price_currency` is EGP and `area_unit` is sqm before comparing prices.
- Do not remove a listing only because its price, area, or price per square meter is unusually high or low.
- Flag unusually low or high values for review and retain them in the dataset.
- Treat the current local IQR flags as preliminary review indicators, not confirmed errors or fraud.

### 5. Price comparison rules
- Compare rental prices only with other rental listings, never with sale listings.
- Build comparisons using similar locations and property characteristics where the data allows.
- Do not create a reference price for a location or group with too few comparable listings.
- If there is not enough comparison data, return `INSUFFICIENT_COMPARISON_DATA` instead of guessing.
- Review possible daily or short-term rentals separately from monthly rentals before using them in a monthly price reference.

### 6. Missing values
- Do not fill missing values automatically during the initial review.
- Keep missing values as missing and record which fields are unavailable.
- Fields that are entirely or mostly missing, such as `rera` or `video_url`, should not be used in the first version of the fraud checks.
- Missing non-essential fields do not automatically make a listing suspicious.
- Missing required fields may produce a `MISSING_REQUIRED_DATA` review reason.

### 7. Duplicates and identifiers
- Keep `listing_id` as the source identifier.
- The initial check found no fully duplicated rows and no repeated `listing_id` values.
- Do not assume that different IDs always mean different properties; cross-listing duplicates require a separate check using URLs, text, seller identifiers, and, if available, image fingerprints.

### 8. Location and dates
- Preserve the original location fields.
- Do not merge or rewrite `city`, `town`, `district`, or `subdistrict` until inconsistent values have been reviewed.
- Use the listing date and data-collection date as separate fields.
- Record the dataset collection period because property prices change over time.

### 9. Privacy and image limitations
- Keep personal contact fields in the protected raw dataset only when access is authorized.
- Exclude emails and phone numbers from routine analysis outputs and backend responses unless the team has a clear need and permission.
- The inspected dataset has an image count but no direct image URL column.
- Do not claim duplicate-image detection is supported by this dataset unless image files or authorized image URLs are obtained.

### 10. Record status values
- `KEEP_FOR_RENT_ANALYSIS`: eligible for the current rental analysis.
- `REVIEW_TYPE_CONFLICT`: listing fields and title/description may disagree.
- `REVIEW_PRICE_ANOMALY`: price is unusual compared with an adequate local comparison group.
- `REVIEW_MISSING_DATA`: a required field is missing.
- `INSUFFICIENT_COMPARISON_DATA`: not enough comparable listings to assess the price.
- `EXCLUDE_FROM_THIS_ANALYSIS`: outside the current scope, such as a sale listing or non-apartment listing; keep it in the raw dataset.
- A listing may have more than one review reason.

### 11. Important limitation
- A review flag means that a listing needs checking; it does not prove fraud.
- Final decisions about holding or rejecting a listing belong to a human administrator.
- This policy is provisional and should be reviewed with the project supervisor and backend team before implementation.

```python
# Create a separate review copy; keep the original DataFrames unchanged
review_df = apartments_df.copy()

# Flag possible sale-related wording in titles and descriptions
sale_terms = r"\bsale\b|for sale|down payment|installment|للبيع|مقدم|تقسيط"

review_df["title_sale_text_flag"] = review_df["title"].str.contains(
    sale_terms,
    case=False,
    na=False
)

review_df["description_sale_text_flag"] = review_df["description"].str.contains(
    sale_terms,
    case=False,
    na=False
)

review_df["possible_type_conflict"] = (
    review_df["title_sale_text_flag"] |
    review_df["description_sale_text_flag"]
)

# Bring in the preliminary local price flags from the earlier analysis
price_flags = rent_price_check_df.set_index("listing_id")[
    "local_price_outlier"
]

comparison_counts = rent_price_check_df.set_index("listing_id")[
    "area_group_count"
]

review_df["local_price_outlier"] = (
    review_df["listing_id"]
    .map(price_flags)
    .fillna(False)
    .astype(bool)
)

review_df["comparison_group_count"] = review_df["listing_id"].map(
    comparison_counts
)

# Mark listings that do not have enough local comparisons
review_df["insufficient_comparison_data"] = (
    review_df["comparison_group_count"].isna() |
    (review_df["comparison_group_count"] < 30)
)

# Build a readable reason for each listing without deleting any rows
def get_review_reasons(row):
    reasons = []

    if row["possible_type_conflict"]:
        reasons.append("CHECK_TITLE_OR_DESCRIPTION")

    if row["local_price_outlier"]:
        reasons.append("UNUSUAL_LOCAL_PRICE")

    if row["insufficient_comparison_data"]:
        reasons.append("INSUFFICIENT_LOCAL_COMPARISONS")

    if not reasons:
        return "No initial review flag"

    return " | ".join(reasons)

review_df["review_reasons"] = review_df.apply(
    get_review_reasons,
    axis=1
)

# Assign a preliminary status for the current analysis
review_df["record_status"] = review_df["review_reasons"].apply(
    lambda reasons: (
        "REVIEW"
        if reasons != "No initial review flag"
        else "KEEP_FOR_RENT_ANALYSIS"
    )
)

# Display the number of records in each status and review reason
print("Record status counts:")
print(review_df["record_status"].value_counts(dropna=False))

print("\nReview reason counts:")
print(review_df["review_reasons"].value_counts(dropna=False))
```

```python
# Create a fresh review copy from the apartment-rental subset
review_df = apartments_df.copy()

# Search for sale-related wording as a review signal, not as proof of a conflict
sale_terms = r"\bsale\b|for sale|down payment|installment|للبيع|مقدم|تقسيط"

review_df["title_text_review_flag"] = review_df["title"].str.contains(
    sale_terms,
    case=False,
    na=False
)

review_df["description_text_review_flag"] = review_df["description"].str.contains(
    sale_terms,
    case=False,
    na=False
)

# Identify listings excluded from the temporary price-check dataset
price_check_ids = set(rent_price_check_df["listing_id"])
review_df["excluded_from_initial_price_check"] = (
    ~review_df["listing_id"].isin(price_check_ids)
)

# Map preliminary price flags and comparison counts back by listing ID
price_flag_map = rent_price_check_df.set_index("listing_id")[
    "local_price_outlier"
]

comparison_count_map = rent_price_check_df.set_index("listing_id")[
    "area_group_count"
]

review_df["local_price_outlier"] = (
    review_df["listing_id"]
    .map(price_flag_map)
    .eq(True)
)

review_df["comparison_group_count"] = review_df["listing_id"].map(
    comparison_count_map
)

# Mark insufficient comparisons only for listings included in the price check
review_df["insufficient_comparison_data"] = (
    ~review_df["excluded_from_initial_price_check"] &
    (
        review_df["comparison_group_count"].isna() |
        (review_df["comparison_group_count"] < 30)
    )
)

# Build separate, readable review reasons for each listing
def build_review_reasons(row):
    reasons = []

    if row["title_text_review_flag"]:
        reasons.append("SALE_RELATED_WORDING_IN_TITLE")

    if row["description_text_review_flag"]:
        reasons.append("SALE_RELATED_WORDING_IN_DESCRIPTION")

    if row["excluded_from_initial_price_check"]:
        reasons.append("EXCLUDED_FROM_TEMPORARY_PRICE_CHECK")

    if row["local_price_outlier"]:
        reasons.append("UNUSUAL_LOCAL_PRICE")

    if row["insufficient_comparison_data"]:
        reasons.append("INSUFFICIENT_LOCAL_COMPARISONS")

    return " | ".join(reasons) if reasons else "No initial review flag"

review_df["review_reasons"] = review_df.apply(
    build_review_reasons,
    axis=1
)

# Assign a preliminary status without deleting any listings
review_df["record_status"] = review_df["review_reasons"].apply(
    lambda reasons: (
        "REVIEW"
        if reasons != "No initial review flag"
        else "KEEP_FOR_RENT_ANALYSIS"
    )
)

# Display status and individual flag counts
print("Record status counts:")
print(review_df["record_status"].value_counts())

print("\nIndividual review-flag counts:")
for flag in [
    "title_text_review_flag",
    "description_text_review_flag",
    "excluded_from_initial_price_check",
    "local_price_outlier",
    "insufficient_comparison_data"
]:
    print(f"{flag}: {review_df[flag].sum()}")
```

```python
import pandas as pd

# Create a full working copy; keep the raw DataFrame unchanged
cleaned_all = df.copy()

# Convert analysis fields to numeric values in the working copy only
for column in ["price_egp", "area_value", "bedrooms", "bathrooms"]:
    cleaned_all[column] = pd.to_numeric(
        cleaned_all[column],
        errors="coerce"
    )

# Identify listings within the current project scope
is_apartment_rental = (
    cleaned_all["property_type"].eq("Apartment") &
    cleaned_all["offering_type"].eq("Residential for Rent") &
    cleaned_all["price_period"].eq("monthly")
)

# Check whether currency and area units are consistent
currency_is_egp = cleaned_all["price_currency"].str.upper().eq("EGP")
area_is_sqm = cleaned_all["area_unit"].str.lower().eq("sqm")

# Check whether price and area are present and positive
valid_price = cleaned_all["price_egp"].notna() & (
    cleaned_all["price_egp"] > 0
)

valid_area = cleaned_all["area_value"].notna() & (
    cleaned_all["area_value"] > 0
)

# Flag clear sale-related wording in the title for manual review
title_sale_flag = cleaned_all["title"].str.contains(
    r"\bsale\b|for sale",
    case=False,
    na=False
)

# Keep description wording as a separate signal only
# Do not exclude a listing based on description keywords alone
description_text_review_flag = cleaned_all["description"].str.contains(
    r"\bsale\b|for sale|down payment|installment|للبيع|مقدم|تقسيط",
    case=False,
    na=False
)

# Add review flags to the working copy
cleaned_all["is_apartment_monthly_rental"] = is_apartment_rental
cleaned_all["title_sale_review_flag"] = title_sale_flag
cleaned_all["description_text_review_flag"] = description_text_review_flag
cleaned_all["invalid_currency_flag"] = ~currency_is_egp
cleaned_all["invalid_area_unit_flag"] = ~area_is_sqm
cleaned_all["invalid_price_flag"] = ~valid_price
cleaned_all["invalid_area_flag"] = ~valid_area

# A listing can enter the preliminary price analysis only if:
# it is a monthly apartment rental, the title has no sale wording,
# the currency is EGP, the area unit is sqm, and price/area are valid
eligible_for_price_analysis = (
    is_apartment_rental &
    ~title_sale_flag &
    currency_is_egp &
    area_is_sqm &
    valid_price &
    valid_area
)

# Create a separate candidate dataset for rental-price analysis
# This is a working subset, not a deletion from the full dataset
rental_price_candidates = cleaned_all.loc[
    eligible_for_price_analysis
].copy()

# Calculate monthly rent per square meter for eligible candidates
rental_price_candidates["monthly_rent_per_sqm"] = (
    rental_price_candidates["price_egp"] /
    rental_price_candidates["area_value"]
)

# Calculate local price ranges using the current city/town fields
group_columns = ["city", "town"]

grouped_prices = rental_price_candidates.groupby(
    group_columns,
    dropna=False
)["monthly_rent_per_sqm"]

rental_price_candidates["comparison_group_count"] = (
    grouped_prices.transform("count")
)

rental_price_candidates["local_q1"] = grouped_prices.transform(
    lambda values: values.quantile(0.25)
)

rental_price_candidates["local_q3"] = grouped_prices.transform(
    lambda values: values.quantile(0.75)
)

rental_price_candidates["local_iqr"] = (
    rental_price_candidates["local_q3"] -
    rental_price_candidates["local_q1"]
)

# Use the preliminary minimum group size for exploratory review only
has_enough_comparisons = (
    rental_price_candidates["comparison_group_count"] >= 30
)

# Flag local IQR outliers only when the comparison group is large enough
rental_price_candidates["local_price_review_flag"] = (
    has_enough_comparisons &
    (
        (
            rental_price_candidates["monthly_rent_per_sqm"] <
            rental_price_candidates["local_q1"] -
            1.5 * rental_price_candidates["local_iqr"]
        ) |
        (
            rental_price_candidates["monthly_rent_per_sqm"] >
            rental_price_candidates["local_q3"] +
            1.5 * rental_price_candidates["local_iqr"]
        )
    )
)

# Identify listings that do not have enough local comparisons
rental_price_candidates["insufficient_comparison_flag"] = (
    ~has_enough_comparisons
)

# Map the price-review results back to the full working copy
price_review_map = rental_price_candidates.set_index(
    "listing_id"
)["local_price_review_flag"]

comparison_count_map = rental_price_candidates.set_index(
    "listing_id"
)["comparison_group_count"]

insufficient_map = rental_price_candidates.set_index(
    "listing_id"
)["insufficient_comparison_flag"]

cleaned_all["local_price_review_flag"] = (
    cleaned_all["listing_id"]
    .map(price_review_map)
    .eq(True)
)

cleaned_all["comparison_group_count"] = cleaned_all["listing_id"].map(
    comparison_count_map
)

cleaned_all["insufficient_comparison_flag"] = (
    cleaned_all["listing_id"]
    .map(insufficient_map)
    .eq(True)
)

# Build readable review reasons without deleting any records
def build_record_reasons(row):
    reasons = []

    if not row["is_apartment_monthly_rental"]:
        reasons.append("OUTSIDE_CURRENT_RENTAL_APARTMENT_SCOPE")

    if row["title_sale_review_flag"]:
        reasons.append("SALE_WORDING_IN_TITLE")

    if row["invalid_currency_flag"]:
        reasons.append("CURRENCY_MISSING_OR_NOT_EGP")

    if row["invalid_area_unit_flag"]:
        reasons.append("AREA_UNIT_MISSING_OR_NOT_SQM")

    if row["invalid_price_flag"]:
        reasons.append("PRICE_MISSING_OR_NOT_POSITIVE")

    if row["invalid_area_flag"]:
        reasons.append("AREA_MISSING_OR_NOT_POSITIVE")

    if row["local_price_review_flag"]:
        reasons.append("UNUSUAL_LOCAL_PRICE_REQUIRES_REVIEW")

    if row["insufficient_comparison_flag"]:
        reasons.append("INSUFFICIENT_LOCAL_COMPARISONS")

    return " | ".join(reasons) if reasons else "NO_INITIAL_REVIEW_FLAG"

cleaned_all["review_reasons"] = cleaned_all.apply(
    build_record_reasons,
    axis=1
)

# Assign a preliminary status to every record
def assign_record_status(row):
    if not row["is_apartment_monthly_rental"]:
        return "OUT_OF_CURRENT_SCOPE"

    if row["invalid_currency_flag"] or row["invalid_area_unit_flag"]:
        return "REVIEW_DATA_FORMAT"

    if row["invalid_price_flag"] or row["invalid_area_flag"]:
        return "REVIEW_MISSING_OR_INVALID_VALUE"

    if row["title_sale_review_flag"]:
        return "REVIEW_POSSIBLE_TYPE_CONFLICT"

    if row["local_price_review_flag"]:
        return "REVIEW_UNUSUAL_LOCAL_PRICE"

    if row["insufficient_comparison_flag"]:
        return "INSUFFICIENT_COMPARISON_DATA"

    return "READY_FOR_INITIAL_PRICE_ANALYSIS"

cleaned_all["record_status"] = cleaned_all.apply(
    assign_record_status,
    axis=1
)

# Create a review queue without personal contact fields
review_columns = [
    "listing_id",
    "title",
    "offering_type",
    "price_period",
    "price_egp",
    "price_currency",
    "area_value",
    "area_unit",
    "city",
    "town",
    "record_status",
    "review_reasons"
]

review_queue = cleaned_all.loc[
    cleaned_all["record_status"].str.startswith("REVIEW")
    | cleaned_all["record_status"].eq("INSUFFICIENT_COMPARISON_DATA"),
    review_columns
].copy()

# Display the results
print("Raw dataset rows:", len(df))
print("Working dataset rows:", len(cleaned_all))
print("Rental price-analysis candidates:", len(rental_price_candidates))

print("\nRecord status counts:")
print(cleaned_all["record_status"].value_counts())

print("\nReview queue rows:", len(review_queue))
```
